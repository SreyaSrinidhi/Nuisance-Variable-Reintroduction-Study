## 3

What the paper is doing (in my words):
It defines what a good visual representation is using statistics: a function $\phi (x)$ of the image data $x$ that (i) keeps all the information needed about the scene $\theta$ for the task (sufficient), (ii) throws away everything irrelevant (minimal), and (iii) is invariant to nuisance transformations $g \in G$ (viewpoint, illumination, etc.).

Key ideas I learned:

* Data vs scene vs statistic.
	* $x$ (and $y$) are images (finite arrays of pixels) — the data.
	* $\theta$ is the scene (geometry, reflectance, illumination, etc.) — generally infinite-dimensional because it’s function-valued.
	* A statistic $T(x)$ (or representation $\phi(x)$) is any function of the data used to reason about $\theta$

* Sufficiency & minimality (why they matter).

  * Sufficient means: once you know $\phi (x)$, the raw pixels $x$ can’t tell you anything more about $\theta$
  * Minimal means: among sufficient stats, $\phi$ contains no extra junk (smallest sigma-algebra / most compressed).

* Nuisances and invariance.
	* Nuisances $g$ are transformations of the data that shouldn’t affect the answer (pose, lighting, small deformations).
	* $\phi$ is $G$-invariant if $\phi (gx)= \phi (x)$ for all $g\in G$. Sensitivity formalizes how much $\phi$ wiggles when $g$ changes.

* How they construct invariants from likelihoods.
	* Treat $p_{\theta}(x)$ as the likelihood (“how likely is image $x$ if the scene is $\theta$?”).
	* Handle nuisances by either profiling (maximize over $g$) or marginalizing (integrate over $g$).
		* The marginal is invariant only with the Haar (“uniform on the group”) measure.
	* In practice, they approximate this with SAL (Sampled Anti-aliased Likelihood): sample a finite set $\{g
i\}$, locally average (anti-alias) around each $g_i$ with a kernel $w$ (mean-pooling), then max-pool across $i$.

* Why classic tricks make sense now.
	* SIFT/DSP-SIFT can be read as (restricted) SAL under translation/scale/contrast nuisances.
	* Pooling ≈ anti-aliasing over nuisances; clamping and joint normalization improve the approximation to the true contrast-invariant likelihood.
	* Receptive fields (local patches) address occlusion/clutter and connect naturally to CNNs.

## 6

What was hard and how I resolved it (or plan to):

* Keeping roles straight $(x,y,θ,g)$.
    * x/y: data (images). $\theta$: scene. $g$: nuisance transformation acting on data (e.g., $y \mapsto gy$).
	* Breakthrough: thinking “for a fixed $\theta$, the function $g \mapsto p \theta (gy)$” is what we sample/smooth over to build invariants.
* Sufficiency as “no information loss.”
	* The formal definition $Law(X∣T)=independent of \theta$ finally clicked: once you know $T$, the leftover detail in $X$ is $\theta$-irrelevant.
	* The info-theory view helped: information content $=H(\theta)−H(\theta∣T)=I(\theta;T)$ (we maximize this, not drive it to 0).
* Likelihood as a (minimal) sufficient statistic.
	* Seeing $L(\theta;x)=p_{\theta}(x)$ as the minimal sufficient statistic clarified why the paper keeps reducing everything to likelihoods + nuisance handling.
* Profile vs marginal likelihood; why Haar.
	* Profile: $sup_g p_{\theta} (gy)$ — invariant because composing by any $h$ just reindexes $g$.
	* Marginal: $\int p_{\theta}(gy)dP(g)$ — invariant only if $dP$ is Haar (left-invariant). That “uniform on the group” idea was new to me.
* Sampling & anti-aliasing over $G$ (not time).
    * The “sampling” is over nuisances $g$ (and over spatial patches), not over $\theta$ and not time-series.
	* Anti-aliasing = local averaging with a kernel $w$ in $G$ before max/stride — exactly the pooling intuition in CNNs/SIFT.
